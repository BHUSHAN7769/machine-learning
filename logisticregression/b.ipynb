```python
#importing dataset and packages
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification

x1, y1 = make_classification(
    n_samples=100,
    n_features=2,
    n_informative=1,
    n_redundant=0,
    n_classes=2,
    n_clusters_per_class=1,
    random_state=41,
    hypercube=False,
    class_sep=50,
)

plt.scatter(x1[:,1], x1[:,0], c=y1, cmap="rainbow")

x = x1
y = y1
x = np.insert(x, 0, 1, axis=1)
y1.shape
```

```python
#defining required
def step(z):
  if z<0:
    return 0
  else:
    return 1
```

```python
#training and creating the core model
wt = np.ones(x.shape[1])
lr = 0.1
for i in range(1,1000):
  j = np.random.randint(0,100)
  y_pred = step(np.dot(x[j], wt))
  wt = wt+lr*(y[j]-y_pred)*x[j]
coef = wt[1:]
intercept = wt[0]
coef,intercept
```

```python
#creating a line
m = -(coef[0]/coef[1])
b = -(intercept/coef[1])

x_values = np.linspace(x1[:, 1].min(), x1[:, 1].max(), 100)
y_values = m*x_values + b
```

```python
#plotting all in graph
plt.scatter(x1[:,1], x1[:,0], c=y1, cmap="rainbow")
plt.plot(x_values,y_values)
```
